```python
import numpy as np
import scipy.constants as const

def estimate_universal_pulse():
    # Constants
    c = const.c
    er = 4.4  # FR4
    e_eff = (er + 1)/2  # Simple approximation for effective permittivity
    
    # From our parameter sweep ranges:
    # Length: 5mm - 50mm
    # Width: ratios 0.8 to 2.0 of length
    
    # Calculate frequency bounds
    f_max = c/(2 * 0.005 * np.sqrt(e_eff))  # Highest f0 (shortest patch)
    f_min = c/(2 * 0.050 * np.sqrt(e_eff))  # Lowest f0 (longest patch)
    
    # Consider up to first higher order mode
    f_highest = f_max * 2
    
    # Design parameters
    fc = np.sqrt(f_min * f_highest)  # Geometric mean for center frequency
    bw = f_highest - f_min  # Required bandwidth
    
    # Pulse parameters
    tau = 1/(np.pi * bw)  # Pulse width for desired bandwidth
    dt = tau/20  # 20 points per pulse width
    t_max = 8 * tau  # 8 times pulse width for good frequency resolution
    
    # Generate time vector and pulse
    t = np.arange(-t_max/2, t_max/2, dt)
    pulse = np.exp(-((t/tau)**2)) * np.cos(2*np.pi*fc*t)
    
    # Calculate and print key metrics
    print(f"Frequency Coverage:")
    print(f"  Lowest f0: {f_min/1e9:.2f} GHz")
    print(f"  Highest f0: {f_max/1e9:.2f} GHz")
    print(f"  Maximum frequency: {f_highest/1e9:.2f} GHz")
    print(f"\nPulse Parameters:")
    print(f"  Center frequency (fc): {fc/1e9:.2f} GHz")
    print(f"  Bandwidth: {bw/1e9:.2f} GHz")
    print(f"  Pulse width (tau): {tau*1e12:.2f} ps")
    print(f"  Time step (dt): {dt*1e12:.2f} ps")
    print(f"  Time window: {t_max*1e9:.2f} ns")
    
    return t, pulse, dt, fc

# Calculate and save the universal pulse
t, pulse, dt, fc = estimate_universal_pulse()

# Save for use in simulations
np.savez('universal_pulse.npz', 
         t=t, 
         pulse=pulse,
         dt=dt,
         fc=fc)

# Plot pulse characteristics
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 6))

# Time domain plot
plt.subplot(121)
plt.plot(t*1e9, pulse)
plt.xlabel('Time (ns)')
plt.ylabel('Amplitude')
plt.title('Pulse Time Domain')
plt.grid(True)

# Frequency domain plot
freq = np.fft.fftfreq(len(t), dt)
spectrum = np.abs(np.fft.fft(pulse))
spectrum_db = 20 * np.log10(spectrum/np.max(spectrum))

plt.subplot(122)
plt.plot(freq/1e9, spectrum_db)
plt.xlabel('Frequency (GHz)')
plt.ylabel('Magnitude (dB)')
plt.title('Pulse Spectrum')
plt.grid(True)
plt.ylim([-40, 5])
plt.xlim([0, 15])

plt.tight_layout()
plt.show()
```
